```python
import pandas as pd
import numpy as np
 
# Varianta 5
data = {
    "wire_lengths": [5670, 8912],
    "client_amount": [120, 310],
    "total_costs": [45000, 78000]
}
df = pd.DataFrame(data)
df.to_csv("centre_regionale.csv", index=False)
 
# Pasul 2.2 — încărcarea datelor din fișierul CSV
df = pd.read_csv("centre_regionale.csv")
 
# Pasul 2.3 — extragerea vectorilor
wire_lengths = df["wire_lengths"].to_numpy()
client_amounts = df["client_amount"].to_numpy()

total_costs = df["total_costs"].to_numpy()
 
print("Vectorul lungimilor de cablu:", wire_lengths)
print("Vectorul numărului de clienți:", client_amounts)
print("Vectorul costurilor totale:", total_costs)

print()
print("Rezolvam ecuatia de tip: Ax = B <=> x = A^(-1) * B")
print()

A = np.matrix([wire_lengths, client_amounts]).T
print("Matricea A:")
print(A)
print()

print("Vectorul B:")
print(total_costs)
print()

invA = A.getI()
print("Matricea inversa A^(-1):")
print(invA)
print()

x = np.matmul(invA, total_costs)
print("Rezultatul ecuatiei (x):", x)

```

```python

# Varianta 14 — compania de construcții

import pandas as pd
import numpy as np

data = {
    "ciment": [2, 3, 1],
    "nisip": [3, 1, 4],
    "otel_beton": [1, 2, 3],
    "total_costs": [4800, 5200, 6100]
}

df = pd.DataFrame(data)
print("Datele introduse:")
print(df)
print()

ciment = df["ciment"].to_numpy()
nisip = df["nisip"].to_numpy()
otel_beton = df["otel_beton"].to_numpy()
total_costs = df["total_costs"].to_numpy()

print("Cantitatea cimentului:", ciment)
print("Cantitatea nisipului:", nisip)
print("Cantitatea oțel-betonului:", otel_beton)
print("Costul total:", total_costs)

print()
print("Rezolvam ecuatia de tip: Ax = B <=> x = A^(-1) * B")
print()

A = np.matrix([ciment, nisip, otel_beton]).T

print("Matricea A:")
print(A)
print()

print("Vectorul B:")
print(total_costs)
print()

invA = A.getI()
print("Matricea inversa A^(-1):")
print(invA)
print()

x = np.matmul(invA, total_costs)
print("Rezultatul ecuatiei (x):", x)

```

```python
# VarIAnta 28 — companIA IT
import pandas as pd
import numpy as np

data = {
    "develop": [0.70, 0.15],
    "infra":   [0.20, 0.60],
    "demand":  [80, 50]
}

df = pd.DataFrame(data)
print("Datele introduse:")
print(df)
print()

develop = df["develop"].to_numpy()
infra   = df["infra"].to_numpy()
demand  = df["demand"].to_numpy()

print("Coeficienții pentru dezvoltare:", develop)
print("Coeficienții pentru infrastructură:", infra)
print("Cererea finală:", demand)

print()
print("Rezolvam ecuatIA de tip: x = (I-A)^-1 * d")
print()

A = np.matrix([develop, infra]).T
print("Matricea A:")
print(A)
print()

I = np.eye(2)

IA = I - A
print("Matricea (I - A):")
print(IA)
print()

invIA = IA.getI()
print("Matricea inversa (I - A)^(-1):")
print(invIA)
print()

x = invIA @ demand
print("Volumul total de activitate necesar pentru fiecare divizie:", x)
```
