```python
import pandas as pd
import numpy as np
 
# Pasul 2.1 — generarea datelor și salvarea într-un fișier CSV
date = {
    "wire_lengths": [5670, 8912],
    "client_amount": [120, 310],
    "total_costs": [45000, 78000]
}
df = pd.DataFrame(date)
df.to_csv("centre_regionale.csv", index=False)
 
# Pasul 2.2 — încărcarea datelor din fișierul CSV
df = pd.read_csv("centre_regionale.csv")
 
# Pasul 2.3 — extragerea vectorilor
wire_lengths = df["wire_lengths"].to_numpy()
client_amounts = df["client_amount"].to_numpy()

total_costs = df["total_costs"].to_numpy()
 
print("Vectorul lungimilor de cablu:", wire_lengths)
print("Vectorul numărului de clienți:", client_amounts)
print("Vectorul costurilor totale:", total_costs)

print()
print("Rezolvam ecuatia de tip: Ax = B <=> x = A^(-1) * B")
print()

A = np.matrix([wire_lengths, client_amounts]).T
print("Matricea A:")
print(A)
print()

print("Vectorul B:")
print(total_costs)
print()

invA = A.getI()
print("Matricea inversa A^(-1):")
print(invA)
print()

x = np.matmul(invA, total_costs)
print("Rezultatul ecuatiei (x):", x)

```
