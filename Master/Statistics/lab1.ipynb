```python
def analyze_dataset(df, key_column, function_columns):
    print("\nChecking for duplicate records:")
    duplicates = df[df.duplicated()]
    if duplicates.empty:
        print("No duplicates found.")
    else:
        print("\nDuplicate records:")
        print(duplicates)

    # Verification whether the columns are function
    for col in function_columns: 
   # Identifying distinct values in the column
        print(f"Identifying distinct values in the column {col}:")
        unique_values = set(df[col]) 
        print(f"Distinct values: {len(unique_values)}") 
        print(sorted(unique_values)) 
        print() 

        # Checking whether the column is a function of the key column
        print(f"Checking whether the column {col} is a function:")
        values_per_key = df.groupby(key_column)[col].nunique() 
        problematic_values = values_per_key[values_per_key > 1] 
        print(f"Matches with multiple values: {len(problematic_values)}") 
        print(problematic_values)
        print() 
# Downloading the file
df = pd.read_csv("matches.csv") 
print(df.head(df.size)) 
print() 

# Displaying data types
print(df.dtypes)
print() 

analyze_dataset(
    df, 
    key_column="match_code", 
    function_columns=["team_main", "team_guest", "competition", "final_score"])
```
