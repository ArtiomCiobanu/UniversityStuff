```python
# Downloading the file
df = pd.read_csv("matches.csv") 
print(df.head()) 
print() 

# Displaying data types
print(df.dtypes)
print() 

# Identifiying unique values in the "competition" column
unique_competitions = set(df["competition"]) 
print(f"Distinct competitions: {len(unique_competitions)}") 
print(sorted(unique_competitions)) 
print() 

# Verification whether the "competition" is a function
competitions_per_match = df.groupby("match_code")["competition"].nunique() 
problematic_competitions = competitions_per_match[competitions_per_match > 1] 
print(f"Teams with multiple competitions: {len(problematic_competitions)}") 
print(problematic_competitions)
print() 


print("\nChecking for duplicate records...")
duplicates = df[df.duplicated()]
if duplicates.empty:
    print("No duplicates found.")
else:
    print("\nDuplicate records:")
    print(duplicates)
```
